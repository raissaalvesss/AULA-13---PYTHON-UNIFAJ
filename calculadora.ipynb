```python
import calc as cl
```

```python
op = input('''Escolha a opção desejada:
            1-Somar
            2-Subtrair
            3-Multiplicar
            4-Dividir
            5-Potência
            6-Raiz Quadrada''')

A = int(input("Entre com o primeiro numero"))
B = int(input("Entre com o segundo numero"))


if op == '1':
    print(cl.somar(A, B))

elif op == '2':
    print(cl.subtrair(A, B))

elif op == '3':
    print(cl.multiplicar(A, B))

elif op == '4':
    print(cl.dividir(A, B))

elif op == '5':
    print(cl.potencia(A, B))

else:
    print(cl.raiz_quadrada(A, B))
```
